```python
class Bmi:
    def __init__(self, weight, height):
        self.weight = weight
        self.height = height
    
    def set_weight(self, weight):
        self.weight = weight
        return self

    def set_height(self, height):
        self.height = height
        return self

    def calculate_bmi(self):
        return self.weight / (self.height ** 2)

    def __str__(self):
        return f"BMI: {self.calculate_bmi():.2f}"
```

```python
bmi = Bmi(70, 1.75)
print(bmi)  # Output: BMI: 22.86
```

```python
bmi.set_height(1.80)
print(bmi)  # Output: BMI: 21.60
```
