```python
# list
myStudent = ["om", "pare", "ole", "beam", "tae", "aum"]
print(type(myStudent))
print(type(myStudent[2]), myStudent[2])

myStudent[2] = "bell"
print(myStudent)

myStudent.insert(3, "ole")
print(len(myStudent), myStudent)

myStudent.pop(2)
print(myStudent)

# myStudent.clear()
# print(len(myStudent))

myStudent.sort(reverse=True)
print(myStudent)

for i in myStudent:
    print(i)

for i in range(len(myStudent)):
    print(myStudent[i])

i = 0
while i < len(myStudent):
    print(myStudent[i])
    i += 1
```

```python
# tuple
myStudent = ("om", "pare", "ole", "beam", "tae", "aum", "ole")

myStudent = list(myStudent)
myStudent[6]="bell"
myStudent = tuple(myStudent)
print(myStudent)

print(myStudent.index("tae"))

(a,b,c,d,e,f,g) = myStudent
print(a,b,c,d,e,f,g)

for i in myStudent:
    print(i)
```

```python
# set
myStudent1 = {"om", "pare", "ole", "beam", "tae"}
myStudent2 = {"aum", "ole", "pare", "tae"}

myStudent3 = myStudent1.union(myStudent2)
print(myStudent3)
myStudent4 = myStudent1.intersection(myStudent2)
print(myStudent4)
myStudent1.intersection_update(myStudent2)
print(myStudent1)
```
