```python

import requests

api = requests.get("https://data.tmd.go.th/api/WeatherToday/V2/?uid=api&ukey=api12345&format=json")
```

```python
json = api.json()
stations = json["Stations"]["Station"]

def check(num):
    if type(num) is dict: 
        return 0
    else:
        return float(num)

def diff(numMax, numMin):
    return round(check(numMax)-check(numMin))

diff2 = lambda numMax, numMin: round(check(numMax)-check(numMin))

for s in stations:
    sta = s["StationNameThai"]
    minTemp = s["Observation"]["MinTemperature"]
    maxTemp = s["Observation"]["MaxTemperature"]
    diffTemp = diff(maxTemp, minTemp)
    diffTemp2 = diff2(maxTemp, minTemp)
    print(sta, diffTemp, diffTemp2)
```
